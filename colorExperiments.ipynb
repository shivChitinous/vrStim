```python
import numpy as np
import scipy as sp
import skimage as ski
import matplotlib.pyplot as plt
```

```python
image = ski.io.imread("/Users/chitniss/Desktop/natural_scene_6.jpeg")
```

```python
color_opponent = image.copy()
blue_image = image.copy()
blue_image[:,:,[0,1]] = 0

p = 10
color_opponent[:,:,1] = np.percentile(image[:,:,1], p)
color_opponent[:,:,0] = 0
```

```python
fig, ax = plt.subplots(1,3,figsize=(10,10))
images = ['full-color','blue only',f'blue + percentile(green, {p})']
ax[0].imshow(image) 
ax[1].imshow(blue_image)
ax[2].imshow(color_opponent)
for i, a in enumerate(ax):
    a.axis('off')
    a.set_aspect('equal')
    a.set_title(images[i], fontsize = 10)
```
